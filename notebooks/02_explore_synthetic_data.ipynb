```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

data = np.load(Path("data/generated/smoke/train.npz"))
ground_truth = data["ground_truth"][0]
dirty = data["dirty"][0]
mask = data["mask"][0]

fig, axes = plt.subplots(3, ground_truth.shape[0], figsize=(3 * ground_truth.shape[0], 8))
for frame_idx in range(ground_truth.shape[0]):
    axes[0, frame_idx].imshow(ground_truth[frame_idx], cmap="inferno", vmin=0.0, vmax=1.0)
    axes[1, frame_idx].imshow(mask[frame_idx], cmap="gray")
    axes[2, frame_idx].imshow(dirty[frame_idx], cmap="inferno")
    axes[0, frame_idx].set_title(f"Frame {frame_idx}")
    for row in range(3):
        axes[row, frame_idx].set_xticks([])
        axes[row, frame_idx].set_yticks([])
axes[0, 0].set_ylabel("Ground truth")
axes[1, 0].set_ylabel("Mask")
axes[2, 0].set_ylabel("Dirty")
plt.tight_layout()
plt.show()
```
