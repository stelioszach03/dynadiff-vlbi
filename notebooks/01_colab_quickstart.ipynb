# dynadiff-vlbi quickstart

This notebook runs a small end-to-end synthetic dynamic VLBI pipeline in Google Colab. It installs the package, generates a toy dataset, trains the compact 3D U-Net for a short run, evaluates classical and learned baselines, and shows one saved reconstruction figure.

```python
import os
from pathlib import Path

candidate_roots = [
    Path.cwd(),
    Path('/content/dynadiff-vlbi'),
    Path('/content/DynaDiff-VLBI'),
]
repo_root = None
for candidate in candidate_roots:
    if (candidate / 'pyproject.toml').exists() and (candidate / 'scripts').exists():
        repo_root = candidate
        break
if repo_root is None:
    raise FileNotFoundError('Could not find the dynadiff-vlbi repository root. Clone or upload the repo first.')
os.chdir(repo_root)
print(f'Using repo root: {repo_root.resolve()}')
```
